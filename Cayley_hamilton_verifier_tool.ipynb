```python
# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import time

# Make NumPy output easier to read
np.set_printoptions(precision=6, suppress=True)

print("Libraries imported successfully.")
```

```python
# ============================================================
# USER INPUT
# ============================================================

# Choose matrix size
matrix_size = 3       # Change to 2 for a 2x2 matrix

# Enter the matrix
# For a 2x2 matrix, use something like:
# [[2, 1],
#  [1, 3]]

# For a 3x3 matrix:
matrix_data = [
    [2, 1, 0],
    [1, 3, 1],
    [0, 1, 2]
]

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

if matrix_size not in [2, 3]:
    raise ValueError("Matrix size must be either 2 or 3.")

A = sp.Matrix(matrix_data)

if A.shape != (matrix_size, matrix_size):
    raise ValueError(
        f"Expected a {matrix_size}x{matrix_size} matrix."
    )

print("Input Matrix A:")
sp.pprint(A)
```

```python
# ============================================================
# MATHEMATICAL MODEL
# CAYLEY-HAMILTON THEOREM
# ============================================================

print("CAYLEY-HAMILTON THEOREM")
print("-" * 65)

print("""
The Cayley-Hamilton theorem states:

Every square matrix satisfies its own characteristic equation.

If

    p(lambda) = det(lambda I - A)

is the characteristic polynomial of A,

then replacing lambda by A gives:

    p(A) = 0

where 0 is the zero matrix.
""")

print("For our matrix:")

I = sp.eye(matrix_size)

lambda_symbol = sp.symbols('lambda')

# Characteristic polynomial
char_poly = A.charpoly(lambda_symbol)

print("\nCharacteristic polynomial:")
sp.pprint(char_poly.as_expr())
```

```python
# ============================================================
# CHARACTERISTIC EQUATION
# ============================================================

print("=" * 65)
print("STEP 1: CHARACTERISTIC EQUATION")
print("=" * 65)

print("\nMatrix A:")
sp.pprint(A)

print("\nIdentity Matrix I:")
sp.pprint(I)

print("\nλI - A:")
lambda_I_minus_A = lambda_symbol * I - A
sp.pprint(lambda_I_minus_A)

print("\nDeterminant det(λI - A):")

det_expression = sp.expand(lambda_I_minus_A.det())
sp.pprint(det_expression)

print("\nCharacteristic Polynomial p(λ):")
sp.pprint(char_poly.as_expr())

print("\nCharacteristic Equation:")
print("p(λ) = 0")
sp.pprint(sp.Eq(char_poly.as_expr(), 0))
```

```python
# ============================================================
# CHARACTERISTIC POLYNOMIAL COEFFICIENTS
# ============================================================

print("=" * 65)
print("CHARACTERISTIC POLYNOMIAL COEFFICIENTS")
print("=" * 65)

coefficients = char_poly.all_coeffs()

print("\nPolynomial:")
sp.pprint(char_poly.as_expr())

print("\nCoefficients:")
for i, coefficient in enumerate(coefficients):
    print(f"Coefficient {i}: {coefficient}")

print("""
For a 3x3 matrix:

p(λ) = λ³ + c₂λ² + c₁λ + c₀

Therefore Cayley-Hamilton gives:

A³ + c₂A² + c₁A + c₀I = 0
""")
```

```python
# ============================================================
# VERIFY CAYLEY-HAMILTON THEOREM
# ============================================================

print("=" * 65)
print("STEP 2: VERIFYING p(A) = 0")
print("=" * 65)

# Calculate p(A)
pA = sp.zeros(matrix_size)

degree = char_poly.degree()

for i, coefficient in enumerate(coefficients):
    power = degree - i
    pA += coefficient * (A ** power)

print("\np(A) =")
sp.pprint(pA)

print("\nIs p(A) exactly the zero matrix?")

if pA == sp.zeros(matrix_size):
    print("YES ✓")
    print("Cayley-Hamilton theorem is VERIFIED.")
else:
    print("NO ✗")
    print("Verification failed.")
```

```python
# ============================================================
# CAYLEY-HAMILTON REDUCTION RULE
# ============================================================

print("=" * 65)
print("CAYLEY-HAMILTON REDUCTION")
print("=" * 65)

print("\nFrom the characteristic equation:")

sp.pprint(sp.Eq(char_poly.as_expr(), 0))

print("\nTherefore, the highest power can be reduced.")

if matrix_size == 2:
    print("""
For a 2x2 matrix:

A² + c₁A + c₀I = 0

Therefore:

A² = -c₁A - c₀I
""")

elif matrix_size == 3:
    print("""
For a 3x3 matrix:

A³ + c₂A² + c₁A + c₀I = 0

Therefore:

A³ = -c₂A² - c₁A - c₀I
""")
```

```python
# ============================================================
# STEP 3: INVERSE USING CAYLEY-HAMILTON
# ============================================================

print("=" * 65)
print("STEP 3: FINDING A^(-1) USING CAYLEY-HAMILTON")
print("=" * 65)

det_A = A.det()

print("\ndet(A) =")
sp.pprint(det_A)

if det_A == 0:

    print("\nA is singular.")
    print("Therefore A^(-1) does not exist.")

else:

    print("\nA is non-singular.")
    print("Therefore A^(-1) exists.")

    if matrix_size == 2:

        # p(λ) = λ² + c1 λ + c0
        c1 = coefficients[1]
        c0 = coefficients[2]

        print("\nFor a 2x2 matrix:")
        print("A² + c₁A + c₀I = 0")

        print("\nMultiplying by A^(-1):")
        print("A + c₁I + c₀A^(-1) = 0")

        A_inverse_CH = -(A + c1 * I) / c0

    else:

        # p(λ) = λ³ + c2 λ² + c1 λ + c0
        c2 = coefficients[1]
        c1 = coefficients[2]
        c0 = coefficients[3]

        print("\nFor a 3x3 matrix:")
        print("A³ + c₂A² + c₁A + c₀I = 0")

        print("\nMultiply by A^(-1):")
        print("A² + c₂A + c₁I + c₀A^(-1) = 0")

        print("\nTherefore:")
        print("A^(-1) = -(A² + c₂A + c₁I) / c₀")

        A_inverse_CH = -(A**2 + c2*A + c1*I) / c0

    print("\nA^(-1) using Cayley-Hamilton:")
    sp.pprint(A_inverse_CH)

    print("\nDirect inverse using SymPy:")
    A_inverse_direct = A.inv()
    sp.pprint(A_inverse_direct)

    print("\nVerification:")
    if sp.simplify(A_inverse_CH - A_inverse_direct) == sp.zeros(matrix_size):
        print("✓ Both inverse calculations agree.")
    else:
        print("✗ Results do not agree.")
```

```python
# ============================================================
# STEP 4: FIND A^8 USING CAYLEY-HAMILTON
# ============================================================

print("=" * 65)
print("STEP 4: FINDING A^8 USING CAYLEY-HAMILTON")
print("=" * 65)


def reduce_power_using_CH(A, n, coefficients):
    """
    Reduce A^n using the characteristic equation.

    The method repeatedly replaces powers greater than or
    equal to the matrix size using Cayley-Hamilton.
    """

    size = A.rows
    I = sp.eye(size)

    # Start with powers A^0, A^1, ..., A^(size-1)
    basis = [I]

    for power in range(1, size):
        basis.append(A**power)

    # If the required power is already in the basis
    if n < size:
        return basis[n]

    # Recurrence:
    #
    # A^size + c_(size-1) A^(size-1)
    # + ... + c0 I = 0
    #
    # Therefore:
    #
    # A^size = -c_(size-1)A^(size-1)
    #           - ... - c0 I

    # Store reduced representations.
    representations = {}

    for power in range(size):
        vector = [0] * size
        vector[power] = 1
        representations[power] = vector

    # Calculate representation for each required power
    for power in range(size, n + 1):

        previous = representations

        # A^power =
        # -c_(size-1) A^(power-1)
        # -c_(size-2) A^(power-2)
        # ...
        # -c0 A^(power-size)

        result = [0] * size

        for j in range(1, size + 1):

            coefficient = coefficients[j]

            lower_power = power - j

            for k in range(size):
                result[k] += -coefficient * previous[lower_power][k]

        representations[power] = result

    # Convert reduced representation into matrix
    final_vector = representations[n]

    result_matrix = sp.zeros(size)

    for i in range(size):
        result_matrix += final_vector[i] * basis[i]

    return sp.simplify(result_matrix)


A8_CH = reduce_power_using_CH(A, 8, coefficients)

print("\nA^8 using Cayley-Hamilton:")
sp.pprint(A8_CH)
```

```python
# ============================================================
# VERIFY A^8
# ============================================================

print("=" * 65)
print("VERIFYING A^8")
print("=" * 65)

A8_direct = A ** 8

print("\nA^8 using Cayley-Hamilton:")
sp.pprint(A8_CH)

print("\nA^8 using direct matrix exponentiation:")
sp.pprint(A8_direct)

difference = sp.simplify(A8_CH - A8_direct)

print("\nDifference:")
sp.pprint(difference)

if difference == sp.zeros(matrix_size):
    print("\n✓ A^8 calculations AGREE.")
else:
    print("\n✗ A^8 calculations do NOT agree.")
```

```python
# ============================================================
# FINAL MATHEMATICAL RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("                 FINAL RESULTS")
print("=" * 70)

print("\n1. INPUT MATRIX A")
sp.pprint(A)

print("\n2. CHARACTERISTIC POLYNOMIAL")
sp.pprint(char_poly.as_expr())

print("\n3. CHARACTERISTIC EQUATION")
sp.pprint(sp.Eq(char_poly.as_expr(), 0))

print("\n4. CAYLEY-HAMILTON VERIFICATION")
sp.pprint(pA)

if pA == sp.zeros(matrix_size):
    print("✓ p(A) = 0 VERIFIED")

print("\n5. DETERMINANT")
sp.pprint(det_A)

if det_A != 0:
    print("\n6. A^(-1) USING CAYLEY-HAMILTON")
    sp.pprint(A_inverse_CH)

    print("\n7. A^8 USING CAYLEY-HAMILTON")
    sp.pprint(A8_CH)

else:
    print("\n6. A^(-1)")
    print("Does not exist because det(A) = 0.")

print("=" * 70)
```

```python
# ============================================================
# BENCHMARK TEST DATASET
# ============================================================

benchmark_matrices = {

    "2x2 Matrix 1": sp.Matrix([
        [2, 1],
        [1, 3]
    ]),

    "2x2 Matrix 2": sp.Matrix([
        [4, 1],
        [2, 3]
    ]),

    "3x3 Matrix 1": sp.Matrix([
        [2, 1, 0],
        [1, 3, 1],
        [0, 1, 2]
    ]),

    "3x3 Matrix 2": sp.Matrix([
        [3, 1, 1],
        [0, 2, 1],
        [1, 0, 4]
    ])
}

print("Benchmark Dataset")
print("-" * 50)

for name, matrix in benchmark_matrices.items():
    print(f"\n{name}:")
    sp.pprint(matrix)
```

```python

# ============================================================
# BENCHMARK: CAYLEY-HAMILTON VS DIRECT COMPUTATION
# ============================================================

def benchmark_matrix(A, repetitions=100):

    coefficients = A.charpoly(lambda_symbol).all_coeffs()

    # ----------------------------
    # Time Cayley-Hamilton method
    # ----------------------------

    start = time.perf_counter()

    for _ in range(repetitions):
        reduce_power_using_CH(A, 8, coefficients)

    end = time.perf_counter()

    ch_time = (end - start) / repetitions

    # ----------------------------
    # Time direct computation
    # ----------------------------

    start = time.perf_counter()

    for _ in range(repetitions):
        A ** 8

    end = time.perf_counter()

    direct_time = (end - start) / repetitions

    return ch_time, direct_time


benchmark_results = []

for name, matrix in benchmark_matrices.items():

    ch_time, direct_time = benchmark_matrix(matrix)

    benchmark_results.append({
        "Matrix": name,
        "Cayley-Hamilton Time (s)": ch_time,
        "Direct Computation Time (s)": direct_time
    })


print("\nBENCHMARK RESULTS")
print("=" * 70)

for result in benchmark_results:

    print(f"\n{result['Matrix']}")
    print(
        f"CH Method   : "
        f"{result['Cayley-Hamilton Time (s)']:.10f} seconds"
    )

    print(
        f"Direct      : "
        f"{result['Direct Computation Time (s)']:.10f} seconds"
    )
```

```python
# ============================================================
# MATRIX INPUT PARSER
# ============================================================

def parse_matrix(matrix_text, size):
    """
    Convert text input into a SymPy matrix.

    Example:
    2 1 0
    1 3 1
    0 1 2
    """

    try:
        rows = matrix_text.strip().split("\n")

        if len(rows) != size:
            raise ValueError(
                f"Please enter exactly {size} rows."
            )

        matrix = []

        for row in rows:
            values = row.split()

            if len(values) != size:
                raise ValueError(
                    f"Each row must contain {size} values."
                )

            matrix.append([
                sp.Rational(value)
                for value in values
            ])

        return sp.Matrix(matrix)

    except Exception as e:
        raise ValueError(
            f"Invalid matrix input: {str(e)}"
        )
```
